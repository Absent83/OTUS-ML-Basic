# Линейная алгебра, математический анализ

```python
import matplotlib.pyplot as plt
import numpy as np

n = 100

A = np.random.randn(n, n)
svd = np.linalg.svd(A)
errors = []

for r in range(1, n + 1):
    A_new = svd.U[:, :r] @ np.diag(svd.S[:r]) @ svd.Vh[:r, :]
    errors.append(np.linalg.norm(A - A_new, 'fro'))

plt.plot(range(1, n + 1), errors)
plt.xlabel('Ранг')
plt.ylabel('Ошибка')
plt.show()
```
