## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
def convert(s: str):
    is_snake_case = s.islower()
    is_pascal_case = s[0].isupper() and s.count('_') == 0

    if is_snake_case:
        return convert_snake_case_to_pascal_case(s)
    elif is_pascal_case:
        return convert_pascal_case_to_snake_case(s)
    else:
        raise Exception("Неизвестный тип строки")


def convert_pascal_case_to_snake_case(s) -> str:
    new_string = ''
    for char in s:
        if char.islower():
            new_string += char
        else:
            if len(new_string) > 0:
                new_string += "_"
            new_string += char.lower()
    return new_string


def convert_snake_case_to_pascal_case(s) -> str:
    new_string = ''
    sub_strings = s.split('_')
    for sub_string in sub_strings:
        new_string += sub_string.title()
    return new_string


string = input("Введите строку: ")
print(convert(string))
```

## Задание 2: Проверка валидности даты

Написать функцию проверяющую валидность введенной даты.

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
def is_valid_date(date_str: str) -> bool:
    ### input format: 29.02.2000

    split_str = date_str.split(".")
    day = int(split_str[0])
    month = int(split_str[1])
    year = int(split_str[2])

    if year < 1:
        return False
    elif not 1 <= month <= 12:
        return False
    elif not 1 <= day <= get_days_in_month(month, year):
        return False
    else:
        return True

def get_days_in_month(month: int, year: int) -> int:
    days_in_months = [31, get_february_days(year), 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    days_in_month = days_in_months[month - 1]
    return days_in_month

def get_february_days(year: int) -> int:
    return 29 if (is_visokosnii(year)) else 28

def is_visokosnii(year: int) -> bool:
    return year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)


input_str = input("Введите дату: ")
print(is_valid_date(input_str))
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
def is_prostoe(n):
    if n <= 1:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


# Ввод с клавиатуры
num = int(input("Введите число: "))
print(is_prostoe(num))
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
def get_name(prompt: str) -> str | None:
    """Запрашивает имя/фамилию, валидирует и возвращает форматированное значение."""
    while True:
        value = input(prompt).strip()

        if value == "":
            return None

        if not value.isalpha():
            print("Ошибка: поле должно содержать только буквы.")
            continue

        return value.capitalize()

def get_age(prompt: str) -> int | None:
    """Запрашивает возраст, валидирует и возвращает целое число."""
    while True:
        value = input(prompt).strip()

        if value == "":
            return None

        if not value.isdigit():
            print("Ошибка: возраст должен быть числом.")
            continue

        age = int(value)
        if not (18 <= age <= 60):
            print("Ошибка: возраст должен быть от 18 до 60.")
            continue

        return age


def get_id(prompt: str) -> str | None:
    """Запрашивает ID, валидирует и возвращает строку из 8 цифр."""
    while True:
        value = input(prompt).strip()

        if value == "":
            return None

        if not value.isdigit():
            print("Ошибка: ID должен содержать только цифры.")
            continue

        return value.zfill(8)


def get_users() -> dict[str, tuple[str, str, int]]:
    users: dict[str, tuple[str, str, int]] = {}

    while True:
        print("\n--- Новый пользователь ---")

        name = get_name("Введите имя: ")
        if name is None:
            break

        surname = get_name("Введите фамилию: ")
        if surname is None:
            break

        age = get_age("Введите возраст: ")
        if age is None:
            break

        # Получаем ID и проверяем уникальность
        while True:
            user_id = get_id("Введите ID: ")
            if user_id is None:
                return users

            if user_id in users:
                print("Ошибка: такой ID уже существует! Введите другой.")
                continue

            break

        users[user_id] = (name, surname, age)

    return users


def print_users_table(users: dict[str, tuple[str, str, int]]):
    print("\n========== СПИСОК ПОЛЬЗОВАТЕЛЕЙ ==========\n")
    print(f"{'ID':<10} {'Имя':<15} {'Фамилия':<15} {'Возраст':<7}")
    print("-" * 50)

    for uid in sorted(users.keys()):
        name, surname, age = users[uid]
        print(f"{uid:<10} {name:<15} {surname:<15} {age:<7}")

    print("\n==========================================\n")



users = get_users()
print_users_table(users)
```
