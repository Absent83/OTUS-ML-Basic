## Задача 2

В копях Мории хоббиты нашли стену, на которой высечены разные натуральные числа. Согласно древним сказаниям, это даты сражений. Хоббиты знают, что сражения происходили только по високосным годам. Помогите хоббитам определить, является ли год с данным числом датой великого сражения. Если это так, то верните строку "YOU SHALL PASS", иначе верните "YOU SHALL NOT PASS". Напомним, что в соответствии с хоббитским календарем, год является високосным, если его номер кратен 4, но не кратен 100, а также если он кратен 400.

```python
def leap_year(year):
    is_leap = year % 4 == 0 and year % 100 != 0 or year % 400 == 0
    if is_leap:
        return 'YOU SHALL PASS'
    else:
        return 'YOU SHALL NOT PASS'


assert leap_year(5) == 'YOU SHALL NOT PASS'
```

Для могущественного магического ритуала Гендальфу необходимо быстро подсчитывать площадь своего амулета, который умеет менять размеры. Известно, что амулет имеет форму треугольника и Гендальф знает длину каждой из сторон. Напишите функцию, которая считает площадь амулета по трем сторонам.  

**Подсказка:** используйте формулу Герона
$$ S = \sqrt{p(p-a)(p-b)(p-c)} $$
$$ p = \frac{a + b + c}{2}$$


```python
def amulet_area(a, b, c):
    p = (a + b + c) / 2
    S = (p * (p - a) * (p - b) * (p - c)) ** 0.5
    # your code here
    return S


assert amulet_area(3, 4, 5) == 6
```

## Задача 4

Хоббиты собираются пешком идти до Мордора и им нужно подсчитать расстояние, которое им предстоит пройти. Хоббиты смогли вспомнить сразу несколько метрик расстояния: евклидово, манхэттена и косинусное, так что ваша задача - написать функцию под каждую из них. Важное условие - используйте только базовые функции для решения.


* Евклидово расстояние
$$ d(a, b) = \sqrt{\sum_i (a_i - b_i)^2} $$
* Расстояние Манхэттена
$$ d(a, b) = \sum_i |a_i - b_i| $$
* Косинусное расстояние
$$ d(a, b) = 1 - \frac{a^\top b}{||a||_2\cdot||b||_2}$$

```python
def cal_euclidean(a, b):
    distance = 0
    for ai, bi in zip(a, b):
        distance = distance + (ai - bi) ** 2
    return distance ** 0.5


def cal_manhattan(a, b):
    distance = 0
    for ai, bi in zip(a, b):
        distance = distance + abs(ai - bi)
    return distance


def cal_cosine(a, b):
    ab_sum = 0
    a2_sum = 0
    b2_sum = 0
    for ai, bi in zip(a, b):
        ab_sum = ab_sum + ai * bi
        a2_sum = a2_sum + ai ** 2
        b2_sum = b2_sum + bi ** 2
    return 1 - ab_sum / ((a2_sum ** 0.5) * (b2_sum ** 0.5))
```

```python
import numpy as np

a = np.random.randint(-10, 10, size=10)
b = np.random.randint(-10, 10, size=10)
print(cal_euclidean(a, b))
print(cal_manhattan(a, b))
print(cal_cosine(a, b))
```

Создайте случайный array (`np.random.rand()`) длинной 100. Преобразуйте его так, чтобы
* Максимальный элемент(ы) был равен 1
* Минимальный элемент(ы) был равен 0
* Остальные элементы в интервале от 0 до 1 остаются прежними

```python
def normalize(a):
    min_val = None
    min_i = []
    max_val = None
    max_i = []
    for i, ai in enumerate(a):
        if ai == min_val:
            min_i.append(i)
        if ai == max_val:
            max_i.append(i)
        if min_val is None or ai < min_val:
            min_val = ai
            min_i = [i]
        if max_val is None or ai > max_val:
            max_val = ai
            max_i = [i]
    for i in min_i:
        a[i] = 0
    for i in max_i:
        a[i] = 1


my_array = np.random.rand(100)
normalize(my_array)
print(np.max(my_array), np.min(my_array))
print(my_array)
```

Создайте array размером $5 \times 6$ с целыми числами в интервале [0,50]. Напечатайте колонку, которая содержит максимальный элемент полученной матрицы

```python
def get_max_column_number(array):
    max_value = None
    max_col = None
    for row_i, row in enumerate(array):
        for col_i, value in enumerate(row):
            if max_value is None or value > max_value:
                max_value = value
                max_col = col_i
    return max_col

def get_column(array, col_number):
    column=[]
    for row in array:
        column.append(int(row[col_number]))
    return column

my_array = np.random.randint(0, 51, size=(5, 6))
print('Shape: ', my_array.shape)
print('Array')
print(my_array)

selected_column_number = get_max_column_number(my_array)
print('Column_number: ', selected_column_number)

selected_column = get_column(my_array, selected_column_number)
print(selected_column)
```

Напишите функцию, которая принимает на вход матрицу (array) X и возвращает все её уникальные строки в виде новой матрицы.

```python
def get_unique_rows(X):
      seen = set()
      unique_rows = []
      for row in X:
          row_tuple = tuple(int(x) for x in row)
          if row_tuple not in seen:
              seen.add(row_tuple)
              unique_rows.append(list(row_tuple))
      return unique_rows

X = np.random.randint(4, 6, size=(10, 3))
print(X)

unique_rows = get_unique_rows(X)
print(unique_rows)
```
